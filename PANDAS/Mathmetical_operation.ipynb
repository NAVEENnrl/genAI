## Mathemitical functions in pandas

#### sum()----sum of all values present in series
#### product()--->product of values in the series

#### mean()---->mean provides us the average values of the data in colum

#### median()----median also provide the idea about the average of data. for coming up with median, we need to sort the value in ascending manner and then the middle value after sorting the data

#### mode()--->provide us the idea that which value is getting repeated maximum number of times.

#### std()---- if i need to get the idea that how the values inside the series are distributed from the avarage or mean value then we can come up with the standard daviation

#### variance---> it also provide the insight about the distribution of the data from themean value and variance is the square of the standard deviation

#### standard_deviation=|mean-v_1|+|mean-v_2|..... total count of values

#### describe()--->describe function provide the idea about the summery of the important mathemetical calculation for the data,
#### get the idea about min,max,mean,std,25%(this mean 25 % of the values out of my all value are less then this specific value),50%,75%,count(total no of rows)
