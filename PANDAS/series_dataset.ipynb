## Series methods

#### 1. head()--->provide the top preview of data as the 5 rows of the dataframe or series, it is quite helpfil with providing the preview of the datset and if we want to preview top 10 rows then we can pass 10 as a argument or for 7 rows we can pass 7 as argument

```python
you_data.category_id.head()
```

```python
you_data.category_id.head(10)
```

```python
you_data.head()
```

#### 2. tail()--- it provide the bottom preview of data as the bottom 5 rows. for eg.. dataset ---> salary column---top 5 rows---salary---7 lpa---->top 10---> top 15 ,,, with tail() bottom 5 rows---salary---4 lpa

```python
you_data.category_id.tail(10)
```

#### 3. sample()--->provide any custom row from the dataset for us to preview it, it provide any random values

```python
you_data.sample(5)
```

#### 4. value_counts()---> it provide the idea about the count of the values(categorical values),

```python
you_data.category_id.value_counts()
```

```python
type(you_data)
```

```python
type(you_data.category_id)
```

#### 5. sort_values()---> this method sorts the values in an ascending manner. it can be useful for us for the numerical column. you can even sort the value in descending manner also, fpr descending pass the ascending =false inside the method

```python
you_data.category_id.sort_values()
```

```python
you_data.category_id.sort_values(ascending=False)
```

#### Method chaining---> we will use multiple method together to get the required values in a way first method output will be second method input and 2nd method output will be the 3rd method input

```python
you_data.category_id.sort_values(ascending=False).head(1)
```

```python
you_data.category_id.sort_values(ascending=False).head(1).values[0]
```

#### inplace---> in order to make a change in our data permanent

```python
you_data.category_id.sort_values(inplace=True)
```
