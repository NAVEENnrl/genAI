## Pandas series is a column of data in a table like in excel or google sheets with labels called as indexes attached to each values

```python
import numpy as np
import pandas as pd
```

```python
pd.Series([40,50,60],index=['Bread','Milk','Butter'],name='Price of Grocessery items')
## or pd.Series(data=[40,50,60],index=['Bread','Milk','Butter'])
## mention data is optional
```

## series is a 1D array that can hold any type of data like integrers, float, String(Objects),boolean

## Creating a series in multiple ways
#### 1. List
#### 2. Numpy array
#### 3. dictionary
#### 4. Custome Index
#### 5. Scaler Value

```python
##Series from List
country=['India','China','USA','England']
pd.Series(country)
```

```python
batsman_runs=[48,64,72,98,114,183]
pd.Series(batsman_runs)
```

```python
subject_name=['math','physics','chemistry','biology','english']
subject_marks=[48,92,64,35,98]
pd.Series(data=subject_marks,index=subject_name,name='subject marks data',dtype='int32')
```

```python
## create series using numpy array
values=np.arange(10,21)
pd.Series(values)
```

#### Create series using dictionary
#### Dictionary is another data structure that stores the data in key value pair and just because it store the data in key value pair
####  so we dont have to explicitly mention the index

```python
data_item={'Math':78,'English':92,'Physics':88,'Chemistry':64}
pd.Series(data_item)
```

```python
values=[1,2,3,4,5]
pd.Series(data=values,index=['a','b','c','d','e'])
```

```python
## Scaler Values inside in series
pd.Series(4,index=['a','b','c','d','e'])
```
