```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("dataset/StudentPerformanceFactors.csv")
print(df)
```

```python
df.head()
```

```python
print(df.columns.tolist())
```

```python
# 1. create a relational plot
sns.relplot(
    data=df,
    x="Exam_Score",
    y="Hours_Studied",
    hue="Gender",
  
)

plt.show()
```

```python
# 2.use hue with a categorical column
sns.relplot(
    data=df,
    x="Exam_Score",
    y="Hours_Studied",
    hue="Family_Income",
)

plt.show()
```

```python
# 3. create the same plot using scatter style
sns.relplot(
    data=df,
    x="Exam_Score",
    y="Hours_Studied",
    hue="Family_Income",
    kind="scatter"
)

plt.show()
```

## Task 2: Line Plot as Scatter & Facet

```python
# 1. Create a line plot using sns.lineplot()
sns.lineplot( 
    data=df,
    x="Exam_Score",
    y="Previous_Scores",
    hue="Gender"
)
plt.title("Previous Score vs Exam Score")
```

```python
# 2. convert the same relationship into a scatter-style line plot
sns.relplot( 
    data=df,
    x="Exam_Score",
    y="Previous_Scores",
    hue="Gender",
    kind="scatter"
)
plt.title("Previous Score vs Exam Score")
```

```python
# 3. use faceting (col or row) to split the plot based on a categorical column
sns.relplot( 
    data=df,
    x="Exam_Score",
    y="Previous_Scores",
    hue="Gender",
    col="Family_Income",
    kind="scatter"

)
plt.title("Previous Score vs Exam Score")
```

## Task 3: Distribution Plots

```python
# 1. Plot histogram using seaborn

sns.histplot(
    data=df,
    x="Exam_Score",
    bins=10
)

plt.title("Distribution of Exam Scores")
plt.show()
```

```python
# 2. Plot KDE Plot

sns.kdeplot(
    data=df,
    x="Exam_Score",
    fill=True
)

plt.title("KDE Plot of Exam Scores")
plt.show()
```

```python
# 3. Plot Rug Plot
sns.rugplot(
    data=df,
    x="Exam_Score"
)

plt.title("Rug Plot of Exam Scores")
plt.show()
```

```python
# 4. Combine Histogram+KDE in a single plot
sns.histplot(
    data=df,
    x="Exam_Score",
    bins=10,
    kde=True
)

plt.title("Histogram and KDE of Exam Scores")
plt.show()
```

## Task 4: Bivariate Distribution Plots

```python
# 1. Create a bivariate histogram
sns.histplot(
    data=df,
    x="Exam_Score",
    y="Hours_Studied",
    bins=15,
    cbar=True
)

plt.title("Bivariate Histogram: Exam vs Hours Studied")
plt.show()
```

```python
# 2. Create a bivariate KDE plot
sns.kdeplot(
    data=df,
    x="Exam_Score",
    y="Hours_Studied",
    fill=True,
    cmap="Blues"
)

plt.title("Bivariate KDE: Exam vs Hours Studied")
plt.show()
```

## Task 5: Matrix Plots

```python
# 1. Create a pair plot using sns.pairplot()
sns.pairplot(
    df,
    vars=["Previous_Scores", "Exam_Score"],
    diag_kind="hist"
)

plt.show()
```

```python
# 2. create a heatmap of correlation matrix
correlation_matrix = df[["Previous_Scores", "Exam_Score"]].corr()

plt.figure(figsize=(6, 4))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Matrix Heatmap")
plt.show()
```

## Task 6: Categorical Plots

```python
# 1. Bar Plot
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Gender",
    y="Exam_Score",
    errorbar=None
)

plt.title("Average Score by Gender")
plt.xticks(rotation=45)
plt.show()
```

```python
# 2. Box Plot
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Gender",
    y="Exam_Score"
)

plt.title("Marks Distribution by Gender")
plt.xticks(rotation=45)
plt.show()
```

```python
# 3. Violin plot

plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="Gender",
    y="Exam_Score"
)

plt.title("Marks Distribution by Gender")
plt.xticks(rotation=45)
plt.show()
```

```python
# 4. Count Plot
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Gender"
)

plt.title("Number of Student by Gender")
plt.xticks(rotation=45)
plt.show()
```

## Task 7. Regression Plots

```python
# 1. Create a regression plot(regplot) between two numerical column
plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="Attendance",
    y="Sleep_Hours",
    scatter_kws={"alpha": 0.6},
    line_kws={"color": "red"}
)

plt.title("Regression Plot: Attendance vs Sleep Hours")
plt.show()
```

```python
# 2. create an implot with hue using a categorical column
sns.lmplot(
    data=df,
    x="Attendance",
    y="Sleep_Hours",
    hue="Gender",
    height=5,
    aspect=1.4
)

plt.title("Regression Plot by Gender")
plt.show()
```

## Task 8: Multi Plots & figure level plots

```python
# 1. Create a FacetGrid
g = sns.FacetGrid(
    data=df,
    col="Gender",
    height=4
)

g.map_dataframe(
    sns.scatterplot,
    x="Attendance",
    y="Exam_Score"
)

g.set_axis_labels("Attendance", "Exam_Score")
g.fig.suptitle("Student Attendance vs Score by Gender", y=1.05)
```

```python
# 1. Create a multi plot dashboard: relplot
sns.relplot(
    data=df,
    x="Attendance",
    y="Exam_Score",
    hue="Gender",
    kind="scatter"
)
```

```python
# catplot:
sns.catplot(
    data=df,
    x="School_Type",
    y="Exam_Score",
    hue="Gender",
    kind="box"
)
```

```python
# displot:
sns.displot(
    data=df,
    x="Exam_Score",
    hue="Gender",
    kind="hist",
    kde=True
)
```
