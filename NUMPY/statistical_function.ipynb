#### statistical function
#### mean()--->syntex for mean()--->np.mean()
#### mean of all values inside the array or matrix


```python
import numpy as np
ary_data=np.round(np.random.random((4,4))*100).astype('int32')
```

```python
ary_data
```

```python
np.mean(ary_data,axis=1)
```

#### std()--> it provides with the standard deviation of all the values of the matrix or array

```python
np.std(ary_data)
```

#### axis=0---column wise std
#### axis=1---row wise std
#### median()--->to calculate median of all values in the array or matrix

```python
np.median(ary_data)
```

```python
np.median(ary_data,axis=0)
```

#### var---> it helps with calculating the varience of all the values inside the array or matrix

```python
np.var(ary_data)
```
