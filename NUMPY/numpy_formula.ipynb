#### axis ----> it provide us with the idea about the max or min or any operation values inside the specific rows and columns or we get the specific value by row wise or column wise

#### axis=0 ----> column wise operation
#### axis=1 ----> row wise operation

```python
ary_data
```

```python
## column wise
np.max(ary_data,axis=0)
```

```python
## row wise
np.max(ary_data,axis=1)
```

#### min()----> it will be used for calculating the minimum value from the matrix or array

```python
np.min(ary_data)
```

```python
np.min(ary_data,axis=0)
```
